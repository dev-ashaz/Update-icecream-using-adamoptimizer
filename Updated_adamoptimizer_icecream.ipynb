```python
import tensorflow as tf
import numpy as np

print("ICE CREAM SALES PREDICTOR")

# 1. Dataset
temperature = np.array([15, 18, 20, 22, 25, 28, 30, 32], dtype=float)
sales = np.array([40, 55, 70, 85, 110, 140, 160, 180], dtype=float)

# 2. Build Model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1])
])

# 3. Compile Model (Using Adam optimizer to prevent NaN gradients)
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
    loss='mean_squared_error'
)

# 4. Train Model
model.fit(temperature, sales, epochs=500, verbose=0)

# 5. Predict
prediction = model.predict(np.array([[35.0]]))
print(f"Predicted Ice Cream Sales at 35°C: {prediction[0][0]:.2f}")
```
